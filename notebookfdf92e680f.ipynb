```python

# ==================================================================================================
# FedFalsify SCSV-PQCR — FROZEN FRESH-DEVELOPMENT KAGGLE RUN
# ONE CELL | KAGGLE CHECKPOINTING | EXACT SOURCE PIN | NO GIT CLONE
#
# WARNING:
# Running this cell SPENDS fresh-development seeds 29101--29105.
# It does NOT touch final-confirmation seeds 11001+.
# The protocol and algorithm are frozen before results are observed.
# ==================================================================================================

import os, sys, json, csv, time, math, shutil, pickle, hashlib, zipfile
import subprocess, platform, urllib.request, traceback
from dataclasses import dataclass, asdict
from pathlib import Path
from statistics import median
from datetime import datetime, timezone
from time import perf_counter

SOURCE_SHA = "e58ff85a93a7498a6220a3032d153c02290de352"
REPO = "AzizulHakim00/fedfalsify"

DEVELOPMENT_SEEDS = (29101, 29102, 29103, 29104, 29105)
FINAL_BLOCK = set(range(11001, 12000))

PROTOCOL_TEXT = r"""# FedFalsify SCSV-PQCR — Frozen Fresh-Development Protocol

**Protocol status:** FROZEN / AUTHORIZED FOR FRESH DEVELOPMENT ONLY  
**Source pin:** `e58ff85a93a7498a6220a3032d153c02290de352`  
**Fresh-development seeds:** `29101, 29102, 29103, 29104, 29105`  
**Final-confirmation namespace:** `11001+` remains untouched  
**Primary successor:** **SCSV-PQCR** (Post-Quarantine Core Refit)  
**Primary comparator:** official **SCSV-ELRC v11** from the pinned source

## 1. Scientific question

Does post-quarantine refitting of the ordinary/shared core improve exact structural recovery and true conditional-mechanism recovery over frozen v11, without increasing conditional false positives, shared/core false positives, runtime, or communication beyond the frozen gates?

This is a fresh-development study, not final confirmation. A GO decision authorizes only a separately frozen independent-validation protocol. A NO-GO permanently spends seeds 29101–29105.

## 2. Frozen method

SCSV-PQCR changes exactly one operational mechanism relative to the v11 certification path:

1. Run the existing v6 high-recall anchor and selector path.
2. Identify ordinary/shared anchor terms and quarantine candidate conditional/exception terms.
3. **Keep the ordinary/shared structure unchanged.**
4. Re-estimate coefficients of that ordinary/shared structure using discovery sufficient-statistic packets after quarantine.
5. Use this post-quarantine refitted core as the frozen reference for v11 effect-localized role inference.
6. Retain v11 weak-heredity provenance.
7. Retain v11 FULL/REDUCED pair invariance.
8. Retain v11 selector/probe held-out conditional evidence.
9. Retain v11 client-median consensus requirement: pooled delta < 0 and client median gain > 0.
10. Retain selector and probe outside-role non-degradation.
11. Retain same-source ambiguity rejection, maximum two operational exceptions, and maximum ten final terms.
12. Final selected structure is refit using the existing final refit path.

No structural core ADD/SWAP repair is permitted.
No consensus-rescue path is permitted.
No positive pooled-delta tolerance is permitted.
No result-dependent threshold change is permitted.
No seed replacement is permitted.

## 3. Frozen benchmark matrix

Use the existing v11 `_scientific_conditions()` matrix with exactly the five fresh seeds above.

Expected:
- 600 matched scientific conditions.
- 2 primary methods per condition.
- 1,200 method rows.

The matrix preserves the existing families, client counts, balance profiles, role profiles, and noise levels from the pinned v11 development harness.

## 4. Primary and safety endpoints

Primary endpoint:
- overall exact structural recovery.

Primary paired effect:
- `PQCR exact recovery - v11 exact recovery`.

Structural safety:
- pooled conditional/deviation precision;
- pooled conditional/deviation recall;
- pooled shared/core precision;
- pooled shared/core recall;
- null-family spurious conditional acceptance;
- null-family spurious shared/core acceptance;
- exact-recovery harm rate: fraction of matched conditions where v11 is exact and PQCR is not.

Robustness:
- true-deviation recovery by main family;
- true-deviation recovery by client count K = 4, 8, 16;
- high-noise main-family recovery;
- dual-role recovery;
- weak-source recovery.

Efficiency:
- median communication ratio PQCR / v11;
- median runtime ratio PQCR / v11.

## 5. Frozen development GO gate

ALL criteria must pass:

A. Exactly 600 matched conditions and exactly seeds 29101–29105.
B. Zero integrity failures / zero unhandled method failures.
C. Overall PQCR exact recovery >= v11 exact recovery + 0.02.
D. Seed-cluster bootstrap 95% CI lower bound for paired exact-rate difference > 0.
E. Pooled conditional/deviation precision >= 0.99.
F. Pooled shared/core precision >= 0.99.
G. Null-family spurious conditional acceptance <= 0.02.
H. Null-family spurious shared/core acceptance <= 0.02.
I. Exact-recovery harm rate <= 0.01.
J. Every main conditional family true-deviation recovery >= 0.92.
K. K=4 true-deviation recovery >= 0.90.
L. K=8 true-deviation recovery >= 0.93.
M. K=16 true-deviation recovery >= 0.95.
N. High-noise main-family true-deviation recovery >= 0.90 AND >= v11 + 0.03.
O. Dual-role true-deviation recovery >= 0.90.
P. Weak-source true-deviation recovery >= 0.85.
Q. Median communication ratio <= 1.05.
R. Median runtime ratio <= 1.25.

No single positive metric can override a failed criterion.

## 6. Statistical analysis

- Report exact recovery for both methods with Wilson 95% intervals.
- Report paired condition counts:
  - PQCR-only success,
  - v11-only success,
  - discordant pairs.
- Report exact two-sided McNemar/binomial p-value as supportive analysis.
- Primary uncertainty for the paired exact-rate difference is a deterministic seed-cluster bootstrap:
  - compute exact-rate difference separately for each of the five seeds;
  - resample the five seed-level differences with replacement;
  - 20,000 replicates;
  - bootstrap RNG seed = 39101;
  - percentile 95% interval.
- Report per-seed exact rates and paired differences.
- Do not remove failed rows. A method exception invalidates criterion B and is retained in the failure log.

## 7. Reproducibility and interruption policy

- Source is downloaded as the exact pinned ZIP; no `git clone`.
- Google Drive is mounted.
- The protocol text is embedded in the notebook and SHA-256 checked before execution.
- Completed matched condition pairs are checkpointed to Drive.
- Resume may skip only condition pairs already fully completed and checksum-valid.
- A partially completed pair is rerun as a pair on the same condition.
- No seed substitution or selective condition deletion.
- CSV, JSON, PKL portable state, plots, protocol, environment manifest, and SHA-256 hashes are saved.

## 8. Claim boundary

If GO:
> Fresh development supports PQCR as a candidate successor to v11 under the frozen synthetic heterogeneous-mechanism matrix. GO does not establish external validity, universal symbolic-regression superiority, catalog-free discovery, privacy, causality, or final Transactions readiness.

If NO-GO:
> PQCR does not advance to final confirmation. Seeds 29101–29105 are permanently spent and must not be used for retuning.

Final-confirmation seeds `11001+` remain untouched under either outcome.
"""
FROZEN_PROTOCOL_SHA256 = "658e0fbb71448cb0e41620f7171d33aabc615d96cd93edb1ac06305c6565e519"

BOOTSTRAP_SEED = 39101
BOOTSTRAP_REPS = 20000

RUN_LABEL = "SCSV_PQCR_FRESH_DEVELOPMENT"
KAGGLE_WORKING = Path("/kaggle/working")
KAGGLE_INPUT = Path("/kaggle/input")

# Optional manual checkpoint path. Leave empty for automatic discovery under /kaggle/input.
# Recommended: upload your Colab pqcr_development_checkpoint.csv as a Kaggle Dataset/Input.
MANUAL_CHECKPOINT_INPUT = ""


# --------------------------------------------------------------------------------------------------
# Helpers
# --------------------------------------------------------------------------------------------------
def banner(title):
    line = "=" * 108
    print("\n" + line)
    print(title)
    print(line, flush=True)

def ok(msg): print(f"✅ {msg}", flush=True)
def info(msg): print(f"ℹ️  {msg}", flush=True)
def warn(msg): print(f"⚠️  {msg}", flush=True)
def fail(msg): raise RuntimeError(msg)

def run_cmd(cmd, cwd=None, timeout=None):
    cmd = [str(x) for x in cmd]
    print("\n$ " + " ".join(cmd), flush=True)
    env = dict(os.environ)
    env["PYTHONUNBUFFERED"] = "1"
    p = subprocess.run(
        cmd,
        cwd=str(cwd) if cwd else None,
        env=env,
        text=True,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        timeout=timeout,
    )
    print(p.stdout, flush=True)
    if p.returncode != 0:
        raise RuntimeError(f"Command failed ({p.returncode}): {' '.join(cmd)}")
    return p.stdout

def download_with_retry(url, target, attempts=4):
    target = Path(target)
    last = None
    for attempt in range(1, attempts + 1):
        try:
            info(f"Download attempt {attempt}/{attempts}")
            req = urllib.request.Request(
                url,
                headers={"User-Agent": "Mozilla/5.0 FedFalsify-PQCR-FreshDev"},
            )
            with urllib.request.urlopen(req, timeout=90) as response, target.open("wb") as f:
                shutil.copyfileobj(response, f)
            if target.stat().st_size < 1000:
                raise RuntimeError("Downloaded archive unexpectedly small.")
            return
        except Exception as exc:
            last = exc
            if target.exists():
                target.unlink()
            warn(f"Download failed: {exc}")
            if attempt < attempts:
                time.sleep(3 * attempt)
    raise RuntimeError("Pinned source download failed.") from last

# ==================================================================================================
# STAGE 0 — FROZEN PROTOCOL / SEED FIREWALL
# ==================================================================================================
banner("STAGE 0/12 — VERIFY FROZEN PROTOCOL BEFORE SPENDING FRESH SEEDS")

observed_protocol_sha = hashlib.sha256(PROTOCOL_TEXT.encode("utf-8")).hexdigest()
assert observed_protocol_sha == FROZEN_PROTOCOL_SHA256, (
    f"Protocol SHA mismatch: {observed_protocol_sha} != {FROZEN_PROTOCOL_SHA256}"
)

assert DEVELOPMENT_SEEDS == (29101, 29102, 29103, 29104, 29105)
assert not (set(DEVELOPMENT_SEEDS) & FINAL_BLOCK)
assert 29001 not in DEVELOPMENT_SEEDS

ok(f"Frozen protocol SHA256: {FROZEN_PROTOCOL_SHA256}")
ok("Fresh development seeds fixed to 29101–29105.")
ok("Final 11001+ namespace remains blocked.")
warn("From this point, a completed scientific run permanently spends 29101–29105.")

# ==================================================================================================
# STAGE 1 — KAGGLE WORKING DIRECTORY + OPTIONAL CHECKPOINT IMPORT
# ==================================================================================================
banner("STAGE 1/12 — KAGGLE WORKING DIRECTORY + PERSISTENT CHECKPOINT")

# Kaggle requirements:
#   Settings -> Internet = ON (needed for the pinned GitHub source ZIP and pip if packages are missing)
#   Accelerator may be GPU, but this NumPy symbolic-recovery workload is CPU-bound.
#
# /kaggle/working persists during the current notebook session and is exported as notebook output.
# To resume in a NEW Kaggle session, attach the previous checkpoint CSV as a Kaggle Dataset/Input.

if not KAGGLE_WORKING.exists():
    fail("This notebook is intended for Kaggle: /kaggle/working was not found.")

STUDY_DIR = KAGGLE_WORKING / RUN_LABEL
STUDY_DIR.mkdir(parents=True, exist_ok=True)

PROTOCOL_PATH = STUDY_DIR / "FROZEN_PQCR_DEVELOPMENT_PROTOCOL.md"
CHECKPOINT_PATH = STUDY_DIR / "pqcr_development_checkpoint.csv"
FAILURE_PATH = STUDY_DIR / "pqcr_development_failure.json"
FINAL_ROWS_PATH = STUDY_DIR / "pqcr_development_rows.csv"

if PROTOCOL_PATH.exists():
    existing = PROTOCOL_PATH.read_text(encoding="utf-8")
    existing_sha = hashlib.sha256(existing.encode("utf-8")).hexdigest()
    if existing_sha != FROZEN_PROTOCOL_SHA256:
        fail("Existing Kaggle working protocol differs from frozen notebook protocol.")
else:
    PROTOCOL_PATH.write_text(PROTOCOL_TEXT, encoding="utf-8")

def _discover_checkpoint_input():
    # 1) Explicit manual path wins.
    if MANUAL_CHECKPOINT_INPUT.strip():
        candidate = Path(MANUAL_CHECKPOINT_INPUT.strip())
        if not candidate.exists():
            fail(f"MANUAL_CHECKPOINT_INPUT does not exist: {candidate}")
        return candidate

    # 2) Automatically find prior checkpoint attached as Kaggle input.
    if KAGGLE_INPUT.exists():
        candidates = sorted(KAGGLE_INPUT.rglob("pqcr_development_checkpoint.csv"))
        if len(candidates) == 1:
            return candidates[0]
        if len(candidates) > 1:
            print("Multiple checkpoint inputs found:")
            for item in candidates:
                print("  •", item)
            fail(
                "Multiple pqcr_development_checkpoint.csv inputs found. "
                "Set MANUAL_CHECKPOINT_INPUT to the exact one you want."
            )
    return None

checkpoint_input = _discover_checkpoint_input()

if checkpoint_input is not None and not CHECKPOINT_PATH.exists():
    shutil.copy2(checkpoint_input, CHECKPOINT_PATH)
    ok(f"Imported prior checkpoint: {checkpoint_input}")

    # If the input dataset also contains the frozen protocol, verify it.
    sibling_protocols = list(checkpoint_input.parent.glob("FROZEN_PQCR_DEVELOPMENT_PROTOCOL.md"))
    if sibling_protocols:
        imported_protocol = sibling_protocols[0].read_text(encoding="utf-8")
        imported_sha = hashlib.sha256(imported_protocol.encode("utf-8")).hexdigest()
        if imported_sha != FROZEN_PROTOCOL_SHA256:
            fail("Imported checkpoint's sibling protocol SHA does not match frozen protocol.")
        ok("Imported checkpoint protocol hash verified.")
    else:
        warn(
            "Checkpoint CSV found without sibling protocol file. "
            "The notebook's own frozen protocol remains authoritative."
        )
elif checkpoint_input is None:
    warn(
        "No prior checkpoint attached. The same frozen study will start from condition 1. "
        "For the cleanest continuation of your interrupted Colab run, attach its checkpoint CSV."
    )

ok(f"Kaggle study directory: {STUDY_DIR}")
ok("Frozen protocol persisted in /kaggle/working.")

# ==================================================================================================
# STAGE 2 — DOWNLOAD / INSTALL EXACT PINNED SOURCE
# ==================================================================================================
banner("STAGE 2/12 — EXACT SOURCE PIN + CURRENT TESTS")

WORK = KAGGLE_WORKING / "fedfalsify_pqcr_fresh"
if WORK.exists():
    shutil.rmtree(WORK)
WORK.mkdir(parents=True)

ZIP_URL = f"https://codeload.github.com/{REPO}/zip/{SOURCE_SHA}"
ZIP_PATH = WORK / "source.zip"
download_with_retry(ZIP_URL, ZIP_PATH)
source_zip_sha = hashlib.sha256(ZIP_PATH.read_bytes()).hexdigest()

with zipfile.ZipFile(ZIP_PATH, "r") as zf:
    bad = zf.testzip()
    if bad is not None:
        fail(f"Corrupt ZIP member: {bad}")
    zf.extractall(WORK)

sources = [p for p in WORK.iterdir() if p.is_dir() and p.name.startswith("fedfalsify-")]
if len(sources) != 1:
    fail(f"Expected one source directory, found {sources}")
SRC = sources[0]

run_cmd([sys.executable, "-m", "pip", "install", "-q", "-e", f"{SRC}[dev]"], timeout=360)
run_cmd([sys.executable, "-m", "pip", "install", "-q", "pandas", "matplotlib", "scipy"], timeout=360)

if str(SRC / "src") not in sys.path:
    sys.path.insert(0, str(SRC / "src"))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import binomtest
from IPython.display import display

test_log = run_cmd(
    [sys.executable, "-m", "pytest", "-q", "tests/test_scsv_v11.py"],
    cwd=SRC,
    timeout=900,
)
(STUDY_DIR / "source_v11_tests.txt").write_text(test_log, encoding="utf-8")
ok("Pinned current v11 tests PASS.")

# ==================================================================================================
# STAGE 3 — IMPORT EXACT PROJECT COMPONENTS
# ==================================================================================================
banner("STAGE 3/12 — IMPORT FROZEN PROJECT APIS")

from fedfalsify.scsv_v6 import scsv_cert_method
from fedfalsify.scsv_v10 import (
    _ordinary_anchor_terms,
    _anchor_exception_terms,
    _conditional_summary,
    _client_gains,
)
from fedfalsify.scsv_v11 import (
    scsv_elrc_v11_method,
    _localize_effect_role,
    _effect_pair,
    _weak_heredity_pass,
)
from fedfalsify.scsv_v8 import _pair_invariant, _outside_safety
from fedfalsify.scsv_v9 import _pooled_delta
from fedfalsify.scsv_diagnostic import _build_packets, _fit_from_packets
from fedfalsify.crossfit_redesign import _ordered_terms, _refit, partition_clients
from fedfalsify.crossfit_surrogate import split_selector_probe
from fedfalsify.scsv_v10_benchmarks import (
    SINGLE_FAMILIES_V10,
    V10_DEVIATIONS,
    generate_v10_benchmark,
    generate_v10_global_test_data,
    v10_catalog,
)
from fedfalsify.scsv_v11_study import (
    DEVELOPMENT_SEEDS as SOURCE_DEVELOPMENT_SEEDS,
    NULL_FAMILIES,
    _scientific_conditions,
)

assert tuple(SOURCE_DEVELOPMENT_SEEDS) == DEVELOPMENT_SEEDS
conditions = tuple(_scientific_conditions(DEVELOPMENT_SEEDS))
assert len(conditions) == 600
assert sorted({int(x[-1]) for x in conditions}) == list(DEVELOPMENT_SEEDS)

ok("Frozen 600-condition development matrix verified.")

# ==================================================================================================
# STAGE 4 — DEFINE MINIMAL PQCR-ONLY SUCCESSOR
# ==================================================================================================
banner("STAGE 4/12 — DEFINE FROZEN SCSV-PQCR")

@dataclass(frozen=True)
class PQCRDiagnostic:
    term: str
    source_term: str
    weak_heredity_passed: bool
    role_admissible: bool
    role_client_ids: tuple
    outside_client_ids: tuple
    deviation_coefficient: float | None
    pair_invariant: bool
    selector_outside_safe: bool
    probe_outside_safe: bool
    pooled_delta: float | None
    client_median_gain: float | None
    positive_pre_ambiguity: bool
    rejection_reason: str

@dataclass(frozen=True)
class PQCROutput:
    method: str
    candidate: object
    anchor_structure: tuple
    ordinary_core_structure: tuple
    core_coefficients: tuple
    candidate_deviations: tuple
    accepted_deviations: tuple
    diagnostics: tuple
    final_structure: tuple
    communication_bytes: int
    runtime_seconds: float
    stop_reason: str

def scsv_pqcr_method(
    datasets,
    catalog,
    *,
    seed,
    target_mse=0.003,
    min_repair_score=0.05,
):
    """
    Frozen PQCR successor:
    identical v11 conditional certification logic except ordinary/shared core
    coefficients are re-estimated on discovery packets after exception quarantine.
    """
    start = perf_counter()

    if int(seed) not in DEVELOPMENT_SEEDS:
        raise ValueError("Fresh PQCR development accepts exactly seeds 29101–29105.")
    if int(seed) in FINAL_BLOCK:
        raise ValueError("Final-confirmation namespace is blocked.")

    anchor = scsv_cert_method(
        datasets,
        catalog,
        seed=seed,
        max_terms=6,
        target_mse=target_mse,
        min_repair_score=min_repair_score,
        use_score_proposer=True,
    )

    anchor_set = set(anchor.selector_structure)
    bank_set = set(anchor.bank.candidate_terms)
    ordinary = tuple(_ordinary_anchor_terms(anchor, catalog))
    anchor_exceptions = tuple(_anchor_exception_terms(anchor, catalog))

    grammar_exceptions = tuple(
        term
        for term in catalog.names()
        if term != "1"
        and catalog.get(term).kind == "exception"
        and catalog.get(term).source_term is not None
        and (
            term in anchor_set
            or catalog.get(term).source_term in bank_set
            or catalog.get(term).source_term in set(ordinary)
        )
    )
    candidates = tuple(dict.fromkeys(anchor_exceptions + grammar_exceptions))

    partitions = partition_clients(datasets, seed=seed, validation_fraction=0.30)
    selectors, probes = split_selector_probe(partitions, seed=seed)

    all_terms = tuple(dict.fromkeys(
        ("1",)
        + tuple(anchor.selector_structure)
        + tuple(anchor.bank.candidate_terms)
        + tuple(candidates)
    ))

    fit_packets, selector_packets, probe_packets, packet_bytes = _build_packets(
        partitions, selectors, probes, catalog, all_terms
    )
    extra_communication = int(packet_bytes)

    # THE ONLY SUCCESSOR MECHANISM:
    # keep ordinary structure fixed, refit ordinary coefficients after quarantine.
    ordinary_structure = _ordered_terms(catalog, set(ordinary) - {"1"})
    core = _fit_from_packets(
        fit_packets,
        all_terms,
        ordinary_structure,
        candidate_id="scsv-pqcr-core-discovery",
    )

    diagnostics = []
    candidate_deviations = []
    positive_terms = []
    positive_sources = {}

    for term in candidates:
        metadata = catalog.get(term)
        source = metadata.source_term
        if source is None:
            continue

        candidate_deviations.append(term)
        weak_ok = _weak_heredity_pass(
            source,
            core,
            anchor.bank.candidate_terms,
        )

        if not weak_ok:
            diagnostics.append(PQCRDiagnostic(
                term=term,
                source_term=source,
                weak_heredity_passed=False,
                role_admissible=False,
                role_client_ids=(),
                outside_client_ids=tuple(str(p.client_id) for p in fit_packets),
                deviation_coefficient=None,
                pair_invariant=False,
                selector_outside_safe=False,
                probe_outside_safe=False,
                pooled_delta=None,
                client_median_gain=None,
                positive_pre_ambiguity=False,
                rejection_reason="SOURCE-PROVENANCE-MISSING",
            ))
            continue

        role, role_bytes = _localize_effect_role(
            partitions,
            core,
            all_terms,
            term,
            catalog,
            seed=seed,
        )
        extra_communication += int(role_bytes)

        if not role.admissible:
            diagnostics.append(PQCRDiagnostic(
                term=term,
                source_term=source,
                weak_heredity_passed=True,
                role_admissible=False,
                role_client_ids=role.role_client_ids,
                outside_client_ids=role.outside_client_ids,
                deviation_coefficient=None,
                pair_invariant=False,
                selector_outside_safe=False,
                probe_outside_safe=False,
                pooled_delta=None,
                client_median_gain=None,
                positive_pre_ambiguity=False,
                rejection_reason="EFFECT-ROLE-NOT-LOCALIZED",
            ))
            continue

        full, reduced, delta = _effect_pair(
            core,
            term,
            tuple(fit_packets[i] for i in role.role_indices),
            all_terms,
            catalog,
            candidate_prefix="scsv-pqcr-discovery",
        )

        invariant = _pair_invariant(full, reduced, term)

        ss, sf, sr, _, _ = _conditional_summary(
            full, reduced, selector_packets, all_terms, role.role_indices
        )
        ps, pf, pr, _, _ = _conditional_summary(
            full, reduced, probe_packets, all_terms, role.role_indices
        )
        pooled = _pooled_delta(ss, sf, sr, ps, pf, pr)

        _, med, _ = _client_gains(
            full,
            reduced,
            selector_packets,
            probe_packets,
            all_terms,
            role.role_indices,
        )

        _, _, selector_safe = _outside_safety(
            full, reduced, selector_packets, all_terms, role.outside_indices
        )
        _, _, probe_safe = _outside_safety(
            full, reduced, probe_packets, all_terms, role.outside_indices
        )

        # EXACT frozen v11-style evidence. NO rescue.
        evidence_ok = bool(
            pooled is not None
            and pooled < 0.0
            and med is not None
            and med > 0.0
        )

        positive = bool(
            invariant
            and selector_safe
            and probe_safe
            and evidence_ok
        )

        if positive:
            positive_terms.append(term)
            positive_sources.setdefault(source, []).append(term)

        reason = (
            "ACCEPTED"
            if positive
            else "PAIR-INVARIANT-FAIL"
            if not invariant
            else "OUTSIDE-NONDEGRADATION-FAIL"
            if not (selector_safe and probe_safe)
            else "POOLED-NOT-SUPPORTED"
            if pooled is None or pooled >= 0.0
            else "CLIENT-MEDIAN-CONTRADICTED"
        )

        diagnostics.append(PQCRDiagnostic(
            term=term,
            source_term=source,
            weak_heredity_passed=True,
            role_admissible=True,
            role_client_ids=role.role_client_ids,
            outside_client_ids=role.outside_client_ids,
            deviation_coefficient=float(delta),
            pair_invariant=bool(invariant),
            selector_outside_safe=bool(selector_safe),
            probe_outside_safe=bool(probe_safe),
            pooled_delta=pooled,
            client_median_gain=med,
            positive_pre_ambiguity=positive,
            rejection_reason=reason,
        ))

    ambiguous_sources = {
        source for source, linked in positive_sources.items()
        if len(linked) > 1
    }
    accepted = tuple(
        term for term in positive_terms
        if catalog.get(term).source_term not in ambiguous_sources
    )

    if len(accepted) > 2:
        accepted = ()

    final_structure = _ordered_terms(
        catalog,
        (set(ordinary_structure) - {"1"}) | set(accepted),
    )
    if len(final_structure) > 10:
        raise RuntimeError("PQCR final structure exceeded ten-term cap.")

    final_candidate, refit_bytes = _refit(
        partitions,
        catalog,
        final_structure,
        include_validation=True,
        candidate_id="scsv-pqcr-final",
    )
    extra_communication += int(refit_bytes)

    return PQCROutput(
        method="scsv-pqcr-fresh-development",
        candidate=final_candidate,
        anchor_structure=tuple(anchor.selector_structure),
        ordinary_core_structure=tuple(ordinary_structure),
        core_coefficients=tuple(float(v) for v in core.coefficients),
        candidate_deviations=tuple(candidate_deviations),
        accepted_deviations=tuple(accepted),
        diagnostics=tuple(diagnostics),
        final_structure=tuple(final_structure),
        communication_bytes=int(anchor.communication_bytes + extra_communication),
        runtime_seconds=float(perf_counter() - start),
        stop_reason=(
            "SCSV-PQCR frozen fresh development; "
            f"anchor={','.join(anchor.selector_structure)}; "
            f"ordinary={','.join(ordinary_structure)}; "
            f"accepted={','.join(accepted)}; "
            f"final={','.join(final_structure)}"
        ),
    )

ok("Frozen PQCR implementation defined.")
ok("No structural repair. No consensus rescue. No tolerance tuning.")

# ==================================================================================================
# STAGE 5 — EVALUATION
# ==================================================================================================
banner("STAGE 5/12 — DEFINE STRUCTURAL METRICS")

def _candidate_terms(candidate):
    return {
        term
        for term, coef in zip(candidate.active_terms, candidate.coefficients)
        if term != "1" and abs(float(coef)) >= 1e-3
    }

def _is_exception(catalog, term):
    return bool(catalog.get(term).kind == "exception")

def evaluate_condition(
    generated,
    candidate,
    method,
    *,
    seed,
    requested_noise,
    runtime_seconds,
    communication_bytes,
    accepted_deviations,
    diagnostics_json,
):
    catalog = v10_catalog()
    predicted = _candidate_terms(candidate)
    target = set(generated.target_terms)

    common = predicted & target
    exact = float(predicted == target)
    term_precision = len(common) / len(predicted) if predicted else float(not target)
    term_recall = len(common) / len(target) if target else 1.0

    true_dev = set(generated.true_deviations)
    pred_dev = predicted & set(V10_DEVIATIONS)
    dev_tp = len(pred_dev & true_dev)
    dev_fp = len(pred_dev - true_dev)
    dev_fn = len(true_dev - pred_dev)

    true_shared = {t for t in target if not _is_exception(catalog, t)}
    pred_shared = {t for t in predicted if not _is_exception(catalog, t)}
    shared_tp = len(true_shared & pred_shared)
    shared_fp = len(pred_shared - true_shared)
    shared_fn = len(true_shared - pred_shared)

    x_test, y_test = generate_v10_global_test_data(generated, seed=seed + 100_000)
    prediction = candidate.predict(x_test, catalog)
    mse = float(np.mean((np.asarray(y_test) - np.asarray(prediction)) ** 2))
    nmse = float(mse / max(float(np.var(y_test)), 1e-12))

    return {
        "family": generated.family,
        "noise_ratio": float(requested_noise),
        "num_clients": int(generated.num_clients),
        "balance_profile": str(generated.balance_profile),
        "role_profile": str(generated.role_profile),
        "seed": int(seed),
        "method": method,
        "exact_recovery": exact,
        "term_precision": float(term_precision),
        "term_recall": float(term_recall),
        "test_nmse": nmse,
        "deviation_tp": int(dev_tp),
        "deviation_fp": int(dev_fp),
        "deviation_fn": int(dev_fn),
        "all_true_deviations_recovered": float(true_dev.issubset(pred_dev)),
        "spurious_deviation": float(dev_fp > 0),
        "shared_tp": int(shared_tp),
        "shared_fp": int(shared_fp),
        "shared_fn": int(shared_fn),
        "spurious_shared": float(shared_fp > 0),
        "runtime_seconds": float(runtime_seconds),
        "communication_bytes": int(communication_bytes),
        "accepted_deviations": ";".join(accepted_deviations),
        "predicted_terms": ";".join(sorted(predicted)),
        "true_terms": ";".join(sorted(target)),
        "diagnostics_json": diagnostics_json,
        "expression": candidate.expression(catalog),
    }

# ==================================================================================================
# STAGE 6 — RESUME / CHECKPOINT INTEGRITY
# ==================================================================================================
banner("STAGE 6/12 — LOAD CHECKPOINT")

condition_columns = [
    "family", "noise_ratio", "num_clients", "balance_profile", "role_profile", "seed"
]
expected_methods = {"v11-elrc", "pqcr"}

if CHECKPOINT_PATH.exists():
    checkpoint = pd.read_csv(CHECKPOINT_PATH)
    info(f"Loaded checkpoint rows: {len(checkpoint)}")
else:
    checkpoint = pd.DataFrame()

def condition_key_from_tuple(item):
    family, num_clients, balance, role, noise, seed = item
    return (
        str(family), float(noise), int(num_clients),
        str(balance), str(role), int(seed)
    )

def row_key(row):
    return (
        str(row["family"]), float(row["noise_ratio"]), int(row["num_clients"]),
        str(row["balance_profile"]), str(row["role_profile"]), int(row["seed"])
    )

completed_keys = set()
if len(checkpoint):
    for key, part in checkpoint.groupby(condition_columns, dropna=False):
        if set(part["method"]) == expected_methods and len(part) == 2:
            completed_keys.add((
                str(key[0]), float(key[1]), int(key[2]),
                str(key[3]), str(key[4]), int(key[5])
            ))

info(f"Fully completed matched condition pairs available for resume: {len(completed_keys)}")

# ==================================================================================================
# STAGE 7 — RUN 600 FRESH MATCHED CONDITIONS
# ==================================================================================================
banner("STAGE 7/12 — RUN FRESH DEVELOPMENT (THIS SPENDS 29101–29105)")

rows = checkpoint.to_dict(orient="records") if len(checkpoint) else []
catalog = v10_catalog()

for index, condition in enumerate(conditions, 1):
    family, num_clients, balance, role, noise, seed = condition
    key = condition_key_from_tuple(condition)

    if key in completed_keys:
        if index % 25 == 0 or index == 600:
            print(f"[{index:03d}/600] RESUME-SKIP completed pair: {family} seed={seed}", flush=True)
        continue

    print(
        f"[{index:03d}/600] {family} | K={num_clients} | "
        f"{balance} | role={role} | noise={noise} | seed={seed}",
        flush=True,
    )

    # If a partial pair exists from an interrupted write, remove it and rerun BOTH methods.
    rows = [r for r in rows if row_key(r) != key]

    try:
        generated = generate_v10_benchmark(
            family,
            nominal_samples_per_client=100,
            noise_ratio=noise,
            seed=seed,
            num_clients=num_clients,
            balance_profile=balance,
            role_profile=role,
        )
        target_mse = max(generated.noise_std**2 * 2.5, 1e-8)

        # Frozen comparator
        out11 = scsv_elrc_v11_method(
            generated.clients,
            catalog,
            seed=seed,
            target_mse=target_mse,
            min_repair_score=0.05,
        )
        rows.append(evaluate_condition(
            generated,
            out11.candidate,
            "v11-elrc",
            seed=seed,
            requested_noise=noise,
            runtime_seconds=out11.runtime_seconds,
            communication_bytes=out11.communication_bytes,
            accepted_deviations=out11.accepted_deviations,
            diagnostics_json=json.dumps(
                [asdict(d) for d in out11.diagnostics],
                sort_keys=True,
                default=str,
            ),
        ))

        # Frozen successor
        outp = scsv_pqcr_method(
            generated.clients,
            catalog,
            seed=seed,
            target_mse=target_mse,
            min_repair_score=0.05,
        )
        rows.append(evaluate_condition(
            generated,
            outp.candidate,
            "pqcr",
            seed=seed,
            requested_noise=noise,
            runtime_seconds=outp.runtime_seconds,
            communication_bytes=outp.communication_bytes,
            accepted_deviations=outp.accepted_deviations,
            diagnostics_json=json.dumps(
                [asdict(d) for d in outp.diagnostics],
                sort_keys=True,
                default=str,
            ),
        ))

        # Atomic-ish checkpoint update after matched pair completion.
        tmp = CHECKPOINT_PATH.with_suffix(".tmp.csv")
        pd.DataFrame(rows).to_csv(tmp, index=False)
        os.replace(tmp, CHECKPOINT_PATH)

        if FAILURE_PATH.exists():
            FAILURE_PATH.unlink()

    except Exception as exc:
        failure = {
            "status": "METHOD-OR-INFRASTRUCTURE-FAILURE",
            "condition_index": index,
            "condition": {
                "family": family,
                "num_clients": num_clients,
                "balance": balance,
                "role": role,
                "noise": noise,
                "seed": seed,
            },
            "exception_type": type(exc).__name__,
            "exception": str(exc),
            "traceback": traceback.format_exc(),
            "timestamp_utc": datetime.now(timezone.utc).isoformat(),
            "protocol_sha256": FROZEN_PROTOCOL_SHA256,
        }
        FAILURE_PATH.write_text(json.dumps(failure, indent=2), encoding="utf-8")
        warn("Run stopped. Failure saved. Do NOT substitute a seed or delete this condition.")
        raise

df = pd.DataFrame(rows)

# Remove any accidental duplicate pair rows by exact full row duplication only; never select by outcome.
df = df.drop_duplicates().reset_index(drop=True)

# Integrity
assert len(df) == 1200, f"Expected 1200 method rows, got {len(df)}"
assert sorted(df.seed.unique().tolist()) == list(DEVELOPMENT_SEEDS)

counts = df.groupby(condition_columns)["method"].agg(list)
assert len(counts) == 600
assert all(set(x) == expected_methods and len(x) == 2 for x in counts)

df.to_csv(FINAL_ROWS_PATH, index=False)
ok("Fresh matrix complete: 600 matched conditions / 1,200 method rows.")

# ==================================================================================================
# STAGE 8 — PRIMARY STATISTICS
# ==================================================================================================
banner("STAGE 8/12 — PRIMARY PAIRED STATISTICS")

key_cols = condition_columns
wide_exact = df.pivot(index=key_cols, columns="method", values="exact_recovery").reset_index()
wide_exact["diff"] = wide_exact["pqcr"] - wide_exact["v11-elrc"]

pqcr_only_success = int(((wide_exact["pqcr"] == 1) & (wide_exact["v11-elrc"] == 0)).sum())
v11_only_success = int(((wide_exact["pqcr"] == 0) & (wide_exact["v11-elrc"] == 1)).sum())
discordant = pqcr_only_success + v11_only_success

mcnemar_p = (
    float(binomtest(min(pqcr_only_success, v11_only_success), n=discordant, p=0.5).pvalue)
    if discordant > 0 else 1.0
)

def wilson_ci(successes, n, z=1.959963984540054):
    if n == 0:
        return (float("nan"), float("nan"))
    phat = successes / n
    denom = 1 + z*z/n
    center = (phat + z*z/(2*n)) / denom
    half = z * math.sqrt((phat*(1-phat)/n) + z*z/(4*n*n)) / denom
    return float(center-half), float(center+half)

method_summary = []
for method in ["v11-elrc", "pqcr"]:
    part = df[df.method == method]
    exact_count = int(part.exact_recovery.sum())
    ci = wilson_ci(exact_count, len(part))
    method_summary.append({
        "method": method,
        "conditions": len(part),
        "exact_recovery": float(part.exact_recovery.mean()),
        "exact_wilson_low": ci[0],
        "exact_wilson_high": ci[1],
        "mean_nmse": float(part.test_nmse.mean()),
        "median_nmse": float(part.test_nmse.median()),
        "median_runtime_seconds": float(part.runtime_seconds.median()),
        "median_communication_bytes": float(part.communication_bytes.median()),
    })
method_summary_df = pd.DataFrame(method_summary)
display(method_summary_df)

seed_rates = (
    df.groupby(["seed", "method"])["exact_recovery"]
    .mean()
    .unstack("method")
    .reset_index()
)
seed_rates["paired_diff"] = seed_rates["pqcr"] - seed_rates["v11-elrc"]
display(seed_rates)

rng = np.random.default_rng(BOOTSTRAP_SEED)
seed_diffs = seed_rates["paired_diff"].to_numpy(dtype=float)
boot = np.empty(BOOTSTRAP_REPS, dtype=float)
for b in range(BOOTSTRAP_REPS):
    sample = rng.choice(seed_diffs, size=len(seed_diffs), replace=True)
    boot[b] = float(np.mean(sample))
boot_low, boot_high = np.percentile(boot, [2.5, 97.5])

print(f"PQCR-only successes: {pqcr_only_success}")
print(f"V11-only successes: {v11_only_success}")
print(f"Discordant pairs: {discordant}")
print(f"Exact two-sided McNemar/binomial p: {mcnemar_p:.8g}")
print(f"Seed-cluster bootstrap 95% CI for exact-rate difference: [{boot_low:.6f}, {boot_high:.6f}]")

# ==================================================================================================
# STAGE 9 — STRUCTURAL / ROBUSTNESS / EFFICIENCY ANALYSIS
# ==================================================================================================
banner("STAGE 9/12 — STRUCTURAL SAFETY + ROBUSTNESS")

def pooled_counts(part, prefix):
    tp = int(part[f"{prefix}_tp"].sum())
    fp = int(part[f"{prefix}_fp"].sum())
    fn = int(part[f"{prefix}_fn"].sum())
    precision = tp / (tp + fp) if tp + fp else 1.0
    recall = tp / (tp + fn) if tp + fn else 1.0
    return tp, fp, fn, float(precision), float(recall)

struct_rows = []
for method in ["v11-elrc", "pqcr"]:
    part = df[df.method == method]
    dtp, dfp, dfn, dp, dr = pooled_counts(part, "deviation")
    stp, sfp, sfn, sp, sr = pooled_counts(part, "shared")
    struct_rows.append({
        "method": method,
        "deviation_tp": dtp,
        "deviation_fp": dfp,
        "deviation_fn": dfn,
        "deviation_precision": dp,
        "deviation_recall": dr,
        "shared_tp": stp,
        "shared_fp": sfp,
        "shared_fn": sfn,
        "shared_precision": sp,
        "shared_recall": sr,
    })
struct_df = pd.DataFrame(struct_rows)
display(struct_df)

pq = df[df.method == "pqcr"].copy()
v11 = df[df.method == "v11-elrc"].copy()

null_pq = pq[pq.family.isin(NULL_FAMILIES)]
null_spurious_dev = float(null_pq.spurious_deviation.mean())
null_spurious_shared = float(null_pq.spurious_shared.mean())

main_pq = pq[pq.family.isin(set(SINGLE_FAMILIES_V10))]
main_v11 = v11[v11.family.isin(set(SINGLE_FAMILIES_V10))]

family_recovery = {
    family: float(
        main_pq[main_pq.family == family].all_true_deviations_recovered.mean()
    )
    for family in SINGLE_FAMILIES_V10
}
client_recovery = {
    str(k): float(
        main_pq[main_pq.num_clients == k].all_true_deviations_recovered.mean()
    )
    for k in (4, 8, 16)
}

high_pq = float(
    main_pq[main_pq.noise_ratio == 0.30].all_true_deviations_recovered.mean()
)
high_v11 = float(
    main_v11[main_v11.noise_ratio == 0.30].all_true_deviations_recovered.mean()
)

dual_recovery = float(
    pq[pq.family == "dual_role_v10"].all_true_deviations_recovered.mean()
)
weak_recovery = float(
    pq[pq.family == "weak_source_role_v10"].all_true_deviations_recovered.mean()
)

paired = wide_exact.copy()
exact_harms = int(((paired["v11-elrc"] == 1) & (paired["pqcr"] == 0)).sum())
harm_rate = exact_harms / 600.0

comm_ratio = float(
    pq.communication_bytes.median() / max(v11.communication_bytes.median(), 1.0)
)
runtime_ratio = float(
    pq.runtime_seconds.median() / max(v11.runtime_seconds.median(), 1e-12)
)

pq_struct = struct_df[struct_df.method == "pqcr"].iloc[0]
overall_v11 = float(v11.exact_recovery.mean())
overall_pq = float(pq.exact_recovery.mean())

gate_values = {
    "v11_exact": overall_v11,
    "pqcr_exact": overall_pq,
    "exact_gain": overall_pq - overall_v11,
    "seed_bootstrap_ci_low": float(boot_low),
    "seed_bootstrap_ci_high": float(boot_high),
    "mcnemar_p": float(mcnemar_p),
    "deviation_precision": float(pq_struct.deviation_precision),
    "deviation_recall": float(pq_struct.deviation_recall),
    "shared_precision": float(pq_struct.shared_precision),
    "shared_recall": float(pq_struct.shared_recall),
    "null_spurious_deviation": null_spurious_dev,
    "null_spurious_shared": null_spurious_shared,
    "exact_harms": exact_harms,
    "exact_harm_rate": float(harm_rate),
    "family_recovery": family_recovery,
    "client_recovery": client_recovery,
    "high_noise_recovery": high_pq,
    "high_noise_v11_recovery": high_v11,
    "dual_role_recovery": dual_recovery,
    "weak_source_recovery": weak_recovery,
    "communication_ratio": comm_ratio,
    "runtime_ratio": runtime_ratio,
}

criteria = {
    "A_integrity_600_conditions_5_seeds": (
        len(wide_exact) == 600
        and sorted(df.seed.unique().tolist()) == list(DEVELOPMENT_SEEDS)
    ),
    "B_zero_method_failures": not FAILURE_PATH.exists(),
    "C_overall_exact_gain_ge_002": overall_pq >= overall_v11 + 0.02,
    "D_seed_bootstrap_ci_low_gt_0": float(boot_low) > 0.0,
    "E_deviation_precision_ge_099": float(pq_struct.deviation_precision) >= 0.99,
    "F_shared_precision_ge_099": float(pq_struct.shared_precision) >= 0.99,
    "G_null_spurious_deviation_le_002": null_spurious_dev <= 0.02,
    "H_null_spurious_shared_le_002": null_spurious_shared <= 0.02,
    "I_exact_harm_rate_le_001": harm_rate <= 0.01,
    "J_each_main_family_recovery_ge_092": min(family_recovery.values()) >= 0.92,
    "K_K4_recovery_ge_090": client_recovery["4"] >= 0.90,
    "L_K8_recovery_ge_093": client_recovery["8"] >= 0.93,
    "M_K16_recovery_ge_095": client_recovery["16"] >= 0.95,
    "N_high_noise_ge_090_and_gain_ge_003": (
        high_pq >= 0.90 and high_pq >= high_v11 + 0.03
    ),
    "O_dual_role_recovery_ge_090": dual_recovery >= 0.90,
    "P_weak_source_recovery_ge_085": weak_recovery >= 0.85,
    "Q_communication_ratio_le_105": comm_ratio <= 1.05,
    "R_runtime_ratio_le_125": runtime_ratio <= 1.25,
}

go = all(criteria.values())
decision = "DEVELOPMENT-GO" if go else "DEVELOPMENT-NO-GO"

banner("FROZEN DEVELOPMENT GATE")
gate_df = pd.DataFrame([
    {"criterion": k, "passed": bool(v)}
    for k, v in criteria.items()
])
display(gate_df)
print(json.dumps(gate_values, indent=2))
print("\nDECISION:", decision)

# ==================================================================================================
# STAGE 10 — PLOTS / FAILURE FORENSICS
# ==================================================================================================
banner("STAGE 10/12 — RESULT FIGURES + FAILURE FORENSICS")

summary_plot = method_summary_df.set_index("method")

for metric in ["exact_recovery", "mean_nmse", "median_runtime_seconds", "median_communication_bytes"]:
    ax = summary_plot[[metric]].plot(kind="bar", legend=False, figsize=(8, 4.2))
    ax.set_title(f"Fresh development — {metric}")
    ax.set_ylabel(metric)
    ax.tick_params(axis="x", rotation=0)
    plt.tight_layout()
    plt.savefig(STUDY_DIR / f"pqcr_dev_{metric}.png", dpi=180, bbox_inches="tight")
    plt.show()
    plt.close()

family_table = (
    pq.groupby(["family", "noise_ratio"])["all_true_deviations_recovered"]
    .mean()
    .reset_index()
)
display(family_table)

# Matched exact failures / repairs
matched_terms = df.pivot(
    index=key_cols,
    columns="method",
    values=["exact_recovery", "predicted_terms", "true_terms", "test_nmse"]
).reset_index()

# ==================================================================================================
# STAGE 11 — SAVE SEALED EVIDENCE
# ==================================================================================================
banner("STAGE 11/12 — SAVE SEALED DEVELOPMENT EVIDENCE")

method_summary_path = STUDY_DIR / "pqcr_development_method_summary.csv"
seed_rates_path = STUDY_DIR / "pqcr_development_seed_rates.csv"
struct_path = STUDY_DIR / "pqcr_development_structural_summary.csv"
family_path = STUDY_DIR / "pqcr_development_family_recovery.csv"
decision_path = STUDY_DIR / "pqcr_development_decision.json"

method_summary_df.to_csv(method_summary_path, index=False)
seed_rates.to_csv(seed_rates_path, index=False)
struct_df.to_csv(struct_path, index=False)
family_table.to_csv(family_path, index=False)

decision_payload = {
    "schema_version": 1,
    "status": decision,
    "protocol_sha256": FROZEN_PROTOCOL_SHA256,
    "source_sha": SOURCE_SHA,
    "development_seeds": list(DEVELOPMENT_SEEDS),
    "conditions": 600,
    "method_rows": 1200,
    "gate_values": gate_values,
    "criteria": criteria,
    "scientific_boundary": (
        "DEVELOPMENT-GO authorizes only a separately frozen independent-validation protocol. "
        "DEVELOPMENT-NO-GO permanently spends 29101–29105. "
        "Final-confirmation 11001+ remains untouched."
    ),
    "timestamp_utc": datetime.now(timezone.utc).isoformat(),
}
decision_path.write_text(json.dumps(decision_payload, indent=2), encoding="utf-8")

portable = {
    "artifact_type": "SCSV-PQCR fresh-development portable evidence",
    "protocol_sha256": FROZEN_PROTOCOL_SHA256,
    "source_sha": SOURCE_SHA,
    "development_seeds": DEVELOPMENT_SEEDS,
    "rows": df.to_dict(orient="records"),
    "method_summary": method_summary_df.to_dict(orient="records"),
    "seed_rates": seed_rates.to_dict(orient="records"),
    "structural_summary": struct_df.to_dict(orient="records"),
    "decision": decision_payload,
}
pkl_path = STUDY_DIR / "pqcr_development_evidence.pkl"
with pkl_path.open("wb") as f:
    pickle.dump(portable, f, protocol=pickle.HIGHEST_PROTOCOL)

with pkl_path.open("rb") as f:
    loaded = pickle.load(f)
assert loaded["protocol_sha256"] == FROZEN_PROTOCOL_SHA256
assert len(loaded["rows"]) == 1200
ok("PKL save/reload PASS.")

manifest = {
    "source_repo": REPO,
    "source_sha": SOURCE_SHA,
    "source_zip_sha256": source_zip_sha,
    "protocol_sha256": FROZEN_PROTOCOL_SHA256,
    "python": sys.version,
    "platform": platform.platform(),
    "numpy": np.__version__,
    "pandas": pd.__version__,
    "scipy": __import__("scipy").__version__,
    "bootstrap_seed": BOOTSTRAP_SEED,
    "bootstrap_reps": BOOTSTRAP_REPS,
    "development_seeds_spent": list(DEVELOPMENT_SEEDS),
    "final_11001_plus_touched": False,
    "timestamp_utc": datetime.now(timezone.utc).isoformat(),
}
manifest_path = STUDY_DIR / "REPRODUCIBILITY_MANIFEST.json"
manifest_path.write_text(json.dumps(manifest, indent=2), encoding="utf-8")

hash_targets = [
    PROTOCOL_PATH,
    FINAL_ROWS_PATH,
    method_summary_path,
    seed_rates_path,
    struct_path,
    family_path,
    decision_path,
    pkl_path,
    manifest_path,
    STUDY_DIR / "source_v11_tests.txt",
]
hashes = {
    p.name: hashlib.sha256(p.read_bytes()).hexdigest()
    for p in hash_targets if p.exists()
}
(STUDY_DIR / "SHA256SUMS.json").write_text(
    json.dumps(hashes, indent=2),
    encoding="utf-8",
)


# Export one downloadable ZIP containing the sealed evidence.
EXPORT_ZIP = KAGGLE_WORKING / "SCSV_PQCR_FRESH_DEVELOPMENT_RESULTS.zip"
if EXPORT_ZIP.exists():
    EXPORT_ZIP.unlink()
shutil.make_archive(
    str(EXPORT_ZIP.with_suffix("")),
    "zip",
    root_dir=STUDY_DIR,
)
ok(f"Kaggle downloadable evidence ZIP created: {EXPORT_ZIP}")

# ==================================================================================================
# STAGE 12 — FINAL LIVE REPORT
# ==================================================================================================
banner("STAGE 12/12 — FINAL FRESH-DEVELOPMENT REPORT")

print(f"Protocol SHA256 : {FROZEN_PROTOCOL_SHA256}")
print(f"Source SHA      : {SOURCE_SHA}")
print(f"Conditions      : 600")
print(f"Method rows     : 1200")
print(f"V11 exact       : {overall_v11:.6f}")
print(f"PQCR exact      : {overall_pq:.6f}")
print(f"Exact gain      : {overall_pq - overall_v11:+.6f}")
print(f"Bootstrap 95% CI: [{boot_low:.6f}, {boot_high:.6f}]")
print(f"McNemar p       : {mcnemar_p:.8g}")
print(f"PQCR dev precision : {float(pq_struct.deviation_precision):.6f}")
print(f"PQCR dev recall    : {float(pq_struct.deviation_recall):.6f}")
print(f"PQCR shared precision: {float(pq_struct.shared_precision):.6f}")
print(f"PQCR shared recall   : {float(pq_struct.shared_recall):.6f}")
print(f"Exact harms      : {exact_harms}")
print(f"Communication ratio: {comm_ratio:.6f}")
print(f"Runtime ratio      : {runtime_ratio:.6f}")
print()
print("FROZEN DECISION:", decision)
print()
print("FINAL CONFIRMATION 11001+ TOUCHED: NO")
print("Kaggle evidence:", STUDY_DIR)
print("Download ZIP  :", EXPORT_ZIP)
print()
if decision == "DEVELOPMENT-GO":
    print("NEXT: freeze a NEW independent-validation protocol. Do not run 11001+ yet.")
else:
    print("NEXT: NO retuning on 29101–29105. Perform failure analysis only on sealed evidence.")
```
